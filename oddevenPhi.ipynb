# $N=2m+1$
-   $0\leq\theta=2p<N$

---



```python
import numpy as np

# Define parameters
m = 3
p = 1
N = 2 * m + 1
theta = 2 * p

# Check if theta is in the valid range [0, N)
if not (0 <= theta < N):
    print(f"Error: theta = {theta} is not in the valid range [0, {N})")
    print("Please choose a different value for p")
    exit()

print(f"Parameters: m = {m}, p = {p}, N = {N}, theta = {theta}")

# Calculate w and G_theta
w = np.exp((-2 * np.pi * 1j) / N)
G_theta = np.zeros((N, N), dtype=complex)
for k in range(N):
    for l in range(N):
        exponent = (k - theta/2) * (l - theta/2)
        G_theta[k, l] = (1/np.sqrt(N)) * (w ** exponent)

# Identity matrix
I = np.eye(N)

# Calculate G_theta squared
G_squared = np.dot(G_theta, G_theta)

# Calculate P_theta and Q_theta
P_theta = 0.5 * (I - G_squared)
Q_theta = 0.5 * (I + G_squared)

# Extract real parts
P_theta_real = np.real(P_theta)
Q_theta_real = np.real(Q_theta)

# Construct Phi matrix with conditional stacking
components = []

# Add q_first (first p+1 rows of Q_theta)
if p + 1 > 0:
    q_first = Q_theta_real[:p+1, :]
    components.append(q_first)
    print(f"Added {q_first.shape[0]} rows from q_first")

# Add q_last (last m-p rows of Q_theta) - only if m-p > 0
if m - p > 0:
    q_last = Q_theta_real[-(m-p):, :]
    components.append(q_last)
    print(f"Added {q_last.shape[0]} rows from q_last")

# Add p_first (first p rows of P_theta)
if p > 0:
    p_first = P_theta_real[:p, :]
    components.append(p_first)
    print(f"Added {p_first.shape[0]} rows from p_first")

# Add p_last (last m-p rows of P_theta) - only if m-p > 0
if m - p > 0:
    p_last = P_theta_real[-(m-p):, :]
    components.append(p_last)
    print(f"Added {p_last.shape[0]} rows from p_last")

# Initialize variables
A = None
B = None
C = None
S = None

# Combine all non-empty components to form Phi
if components:
    Phi = np.vstack(components)
    print(f"Phi shape: {Phi.shape}")

    # Check if Phi is square and invertible
    if Phi.shape[0] == Phi.shape[1]:
        try:
            # Calculate C = Re(G_theta) and S = imag(G_theta)
            C = np.real(G_theta)
            S = np.imag(G_theta)

            # Calculate inverse of Phi
            Phi_inv = np.linalg.inv(Phi)

            # Calculate the transformation: Phi @ C @ Phi_inv
            result_1 = Phi @ C @ Phi_inv
            result_2 = Phi @ S @ Phi_inv

            print("\nResult_1 = Phi @ C @ Phi_inv:")
            print(np.array_str(result_1, precision=6, suppress_small=True))

            print("\nResult_2 = Phi @ S @ Phi_inv:")
            print(np.array_str(result_2, precision=6, suppress_small=True))

            # Check if we can extract the A
            if result_1.shape[0] >= m + 1:
                A = result_1[0:m+1, 0:m+1]
                print("\nsquare A:")
                print(np.array_str(A @ A, precision=6, suppress_small=True))
            else:
                print(f"\nCannot extract A of size {m+1}x{m+1} from result shape {result_1.shape}")

            # Check if we can extract the B
            if result_2.shape[0] >= m + 1:
                B = result_2[m+1:, m+1:]
                print("\nsquare B:")
                print(np.array_str(B @ B, precision=6, suppress_small=True))
            else:
                print(f"\nCannot extract B of size {m}x{m} from result shape {result_2.shape}")

        except np.linalg.LinAlgError:
            print("\nError: Phi matrix is singular and cannot be inverted")
        except Exception as e:
            print(f"\nError: {e}")
    else:
        print(f"\nError: Phi is not square ({Phi.shape}), cannot invert")
else:
    print("\nError: No components to form Phi matrix")

# Print results
if C is not None:
    print("\nC:")
    print(np.array_str(C, precision=6, suppress_small=True))

if S is not None:
    print("\nS:")
    print(np.array_str(S, precision=6, suppress_small=True))

if A is not None:
    print("\nA:")
    print(np.array_str(A, precision=6, suppress_small=True))
```
