```python
import numpy as np

def generate_G1_matrix(N, t):
    w = np.exp(-2j * np.pi / N)
    G1 = np.zeros((N, N), dtype=complex)
    for k in range(N):
        for l in range(N):
            exponent = (k + t / 2) * (l + t / 2)
            G1[k, l] = (1 / np.sqrt(N)) * (w ** exponent)
    return G1

def generate_G2_matrix(N, t):
    w = np.exp(-2j * np.pi / N)
    G2 = np.zeros((N, N), dtype=complex)
    shift = (t + 2 * N) / 2
    for k in range(N):
        for l in range(N):
            exponent = (k + shift) * (l + shift)
            G2[k, l] = (1 / np.sqrt(N)) * (w ** exponent)
    return G2

def compare_G1_G2(N, t, tol=1e-12):
    G1 = generate_G1_matrix(N, t)
    G2 = generate_G2_matrix(N, t)
    diff = np.max(np.abs(G1 - G2))
    return diff < tol

# Main loop for t in range
N = 10
t_min = -50
t_max = 50

for t in range(t_min, t_max + 1):
    if compare_G1_G2(N, t):
        print(f"YES (t = {t})")
    else:
        print(f"NO  (t = {t})")
```
