G^4


```python
import numpy as np

result = 'Positive'

for N in range(10, 100, 5):
    for t in range(-30, 30, 4):
        w = np.exp((-2j * np.pi) / N)
        G = np.empty((N, N), dtype=complex)
        for k in range(N):
            for l in range(N):
                G[k, l] = (1/np.sqrt(N)) * (w ** ((k + t/2) * (l + t/2)))
        G_power_4 = np.linalg.matrix_power(G, 4)

        if not np.allclose(G_power_4, np.eye(N), atol=1e-2):
            print(f"For N = {N}, t = {t}, something is wrong")
            result = 'Negative'
            break
    if result == 'Negative':
        break

if result == 'Positive':
    print('All raised to power 4 become the identity matrix. Okay, keep going')
```

```python
import numpy as np

N = 8

for t in list(range(-30, 0)) + list(range(1, 31)):
    w = np.exp((-2j * np.pi) / N)
    G = np.empty((N, N), dtype=complex)
    for k in range(N):
        for l in range(N):
            G[k, l] = (1/np.sqrt(N)) * (w ** ((k + t/2) * (l + t/2)))
            G[k, l] = G[k, l] ** 2

    G_rounded = np.round(G, decimals=2)

    print(f"\n---------- Result for t = {t} (Rounded) ----------")
    print(G_rounded)
```
