```python
import numpy as np

# ----------------------------
# Parameters
# ----------------------------
m = 2
N = 4*m + 2     # N = 10
theta = 0

# ----------------------------
# Compute w and G
# ----------------------------
w = np.exp(-2j * np.pi / N)

G = np.zeros((N, N), dtype=complex)

for k in range(N):
    for l in range(N):
        exponent = (k - theta/2) * (l - theta/2)
        G[k, l] = (1 / np.sqrt(N)) * w ** exponent

# ----------------------------
# DCT and DST
# ----------------------------
DCT = np.real(G)
DST = np.imag(G)

# Squares
DCT2 = DCT @ DCT
DST2 = DST @ DST

# ----------------------------
# P matrices
# ----------------------------
P_1  = 0.25 * (DCT2 + DCT)
P_m1 = 0.25 * (DCT2 - DCT)
P_i  = 0.25 * (DST2 - DST)
P_mi = 0.25 * (DST2 + DST)

# ----------------------------
# Extract X, Y from P_1, P_m1 (columns 0...m)
# ----------------------------
X = P_1[:, 0:m+1]
Y = P_m1[:, 0:m+1]

# ----------------------------
# Extract Z, T from P_i, P_mi (columns 1...m)
# ----------------------------
Z = P_i[:, 1:m+1]
T = P_mi[:, 1:m+1]

# ----------------------------
# Build V = [X | Y | Z | T]
# ----------------------------
V = np.hstack([X, Y, Z, T])
print(V.shape)
```

```python
import numpy as np

# ----------------------------
# Parameters
# ----------------------------
N = 10        # Example
m = 2         # Example, can be changed

# ----------------------------
# Index arrays
# ----------------------------
k = np.arange(N).reshape(N,1)  # column vector
l = np.arange(N).reshape(1,N)  # row vector

# ----------------------------
# DCT and DST matrices
# ----------------------------
DCT = (1/np.sqrt(N)) * np.cos(2*np.pi*k*l/N)
DST = (1/np.sqrt(N)) * np.sin(2*np.pi*k*l/N)

# ----------------------------
# Kronecker delta matrices
# δ_0, δ_l, δ_{N-l} for all l
# ----------------------------
delta_0   = np.zeros((N,N))
delta_0[0,:] = 1                 # δ_{k,0} for all l
delta_l   = np.eye(N)             # δ_{k,l} (identity)
delta_N_l = np.fliplr(np.eye(N))  # δ_{k,N-l}

# ----------------------------
# P matrices
# ----------------------------
P1   = DCT + 0.5*(delta_0 + delta_l + delta_N_l)
P_m1 = DCT - 0.5*(delta_0 + delta_l + delta_N_l)

P_i   = DST - 0.5*(delta_l - delta_N_l)
P_mi  = DST + 0.5*(delta_l - delta_N_l)

# ----------------------------
# Extract X, Y, Z, T
# ----------------------------
X = P1[:, 0:m+1]       # columns 0..m
Y = P_m1[:, 0:m+1]     # columns 0..m
Z = P_i[:, 1:m+1]      # columns 1..m
T = P_mi[:, 1:m+1]     # columns 1..m

# ----------------------------
# Build V = [X Y Z T]
# ----------------------------
V = np.hstack([X, Y, Z, T])

# ----------------------------
# Display
# ----------------------------
np.set_printoptions(precision=4, suppress=True)
print("X.shape =", X.shape)
print("Y.shape =", Y.shape)
print("Z.shape =", Z.shape)
print("T.shape =", T.shape)
print("V.shape =", V.shape)
```

```python
import numpy as np

# ----------------------------
# Parameters
# ----------------------------
m = 2
N = 4*m + 2
p = 0
t = 2*p
w = np.exp(-2j * np.pi / N)  # complex exponential

# ----------------------------
# Construct Gt
# ----------------------------
Gt = np.zeros((N, N), dtype=complex)
for k in range(N):
    for l in range(N):
        Gt[k, l] = (1/np.sqrt(N)) * w**((k - t/2) * (l - t/2))

# ----------------------------
# Real and Imag parts
# ----------------------------
C = np.real(Gt)
S = np.imag(Gt)

# ----------------------------
# Construct P and Ph
# ----------------------------
P = 0.5 * np.eye(N)
P[0,0] = 1
P[2*m+1, 2*m+1] = 1

for k in range(1, 2*m+1):
    P[k, N - k] = 0.5
    P[k + 2*m + 1, N - k - 2*m - 1] = -0.5

Ph = P

# ----------------------------
# V and W matrices
# ----------------------------
Ph_inv = np.linalg.inv(Ph)
V = Ph @ C @ Ph_inv
W = Ph @ S @ Ph_inv

# ----------------------------
# Split into A and B
# ----------------------------
A = V[0:2*m+2, 0:2*m+2]
B = W[2*m+2:N, 2*m+2:N]

# ----------------------------
# Construct X, Y, Z, T
# ----------------------------
X = A + np.eye(2*m+2)
Y = A - np.eye(2*m+2)
Z = B + np.eye(2*m)
T = B - np.eye(2*m)

# ----------------------------
# Take submatrices
# ----------------------------
X1 = X[:, 0:m+1]
Y1 = Y[:, 0:m+1]
Z1 = Z[:, 0:m]
T1 = T[:, 0:m]

# ----------------------------
# Add zeros as in MATLAB
# ----------------------------
X2 = np.vstack([X1, np.zeros((m+2, m+1))])
Y2 = np.vstack([Y1, np.zeros((m+2, m+1))])
Z2 = np.vstack([np.zeros((2*m+2, m)), Z1])
T2 = np.vstack([np.zeros((2*m+2, m)), T1])

# ----------------------------
# Apply Ph inverse
# ----------------------------
X3 = Ph_inv @ X2
Y3 = Ph_inv @ Y2
Z3 = Ph_inv @ Z2
T3 = Ph_inv @ T2

# ----------------------------
# Construct final V
# ----------------------------
V_final = np.hstack([X3, Y3, Z3, T3])

# ----------------------------
# Display size of V
# ----------------------------
print("V_final.shape =", V_final.shape)
```
