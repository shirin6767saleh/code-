```python
import numpy as np

# --------------------------
# --- First V ---
# --------------------------
m1 = 1
N1 = 4*m1 + 2
p1 = 0
t1 = 2*p1
w1 = np.exp(-2*np.pi*1j/N1)

Gt1 = np.zeros((N1, N1), dtype=complex)
for k in range(N1):
    for l in range(N1):
        Gt1[k, l] = (1/np.sqrt(N1)) * w1**((k - t1/2)*(l - t1/2))

C1 = np.real(Gt1)
S1 = np.imag(Gt1)

P1 = 0.5*np.eye(N1)
P1[0,0] = 1
P1[2*m1+1, 2*m1+1] = 1
for k in range(1, 2*m1+1):
    P1[k, N1-k-1] = 0.5
    P1[k+2*m1+1, N1-k-2*m1-1] = -0.5

Ph1 = P1
V1 = Ph1 @ C1 @ np.linalg.inv(Ph1)

# --------------------------
# --- Second V ---
# --------------------------
m2 = 1
N2 = 4*m2
theta2 = N2-1
w2 = np.exp(-2*np.pi*1j/N2)

G2 = np.zeros((N2, N2), dtype=complex)
for k in range(N2):
    for l in range(N2):
        G2[k,l] = (1/np.sqrt(N2)) * w2**((k - theta2/2)*(l - theta2/2))

I2 = np.eye(N2)
G2_squared = G2 @ G2
P2 = 0.5*(I2 + G2_squared)
Q2 = 0.5*(I2 - G2_squared)
Phi2 = np.vstack([P2[:2*m2,:], Q2[2*m2:,:]])
V2 = Phi2 @ np.real(G2) @ np.linalg.inv(Phi2)

# --------------------------
# --- Third V ---
# --------------------------
m3 = 1
N3 = 4*m3
theta3 = 2*N3-1
w3 = np.exp(-2*np.pi*1j/N3)

G3 = np.zeros((N3, N3), dtype=complex)
for k in range(N3):
    for l in range(N3):
        G3[k,l] = (1/np.sqrt(N3)) * w3**((k - theta3/2)*(l - theta3/2))

I3 = np.eye(N3)
G3_squared = G3 @ G3
P3 = 0.5*(I3 + G3_squared)
Q3 = 0.5*(I3 - G3_squared)
Phi3 = np.vstack([P3[:2*m3,:], Q3[2*m3:,:]])
V3 = Phi3 @ np.real(G3) @ np.linalg.inv(Phi3)

# --------------------------
# --- Compare spans ---
# --------------------------
def compare_span(Va, Vb, tol=1e-10):
    if Va.shape[0] != Vb.shape[0]:
        return None  # Cannot compare different row sizes
    combined = np.hstack([Va, Vb])
    rank_diff = np.linalg.matrix_rank(combined, tol) - Va.shape[1]
    return rank_diff <= tol

V_list = [V1, V2, V3]
N_list = [N1, N2, N3]
names = ['V1','V2','V3']

for i in range(3):
    for j in range(i+1, 3):
        if N_list[i] != N_list[j]:
            print(f"Cannot compare {names[i]} and {names[j]}: different number of rows ({N_list[i]} vs {N_list[j]})")
        else:
            if compare_span(V_list[i], V_list[j]):
                print(f"{names[i]} and {names[j]} span the same space.")
            else:
                print(f"{names[i]} and {names[j]} do NOT span the same space.")

# --------------------------
# Print V1, V2, V3
# --------------------------
print("V1 =\n", V1)
print("V2 =\n", V2)
print("V3 =\n", V3)
```
