#$\theta=0$

```python
import numpy as np
from scipy.linalg import block_diag

m = 2
N = 4 * m
k = np.arange(N).reshape(N, 1)
l = np.arange(N).reshape(1, N)
w = np.exp(-2j * np.pi / N)
F = (1 / np.sqrt(N)) * (w ** (k * l))


t = 0
w = np.exp((-2j * np.pi) / N)

# Cosine & Sine matrices
Ct = np.zeros((N, N))
St = np.zeros((N, N))
for k_idx in range(N):
    for l_idx in range(N):
        angle = (k_idx - t/2) * (l_idx - t/2) * (2 * np.pi / N)
        Ct[k_idx, l_idx] = (1 / np.sqrt(N)) * np.cos(angle)
        St[k_idx, l_idx] = (1 / np.sqrt(N)) * np.sin(angle)

# J, I, X, J_N
J = np.flipud(np.eye(N-1))
I = np.eye(N-1)
X = 0.5 * np.eye(N-1) + 0.5 * np.fliplr(np.eye(N-1))
J_N = block_diag(1, X)

X1 = 0.5 * np.eye(N-1) - 0.5 * np.fliplr(np.eye(N-1))
J_N1 = block_diag(0, X1)

# Matrices
P1 = Ct + J_N
Pm1 = J_N - Ct
Pt = St - J_N1
Pt1 = St + J_N1

# Slices
slice1 = P1[:, 0:m]
slice2 = P1[:, 2*m].reshape(-1, 1)
slice3 = Pm1[:, 0:m]
slice4 = Pt[:, 1:m]
slice5 = Pt1[:, 1:m]
slice6 = Pt1[:, 2*m-1].reshape(-1, 1)

A = np.concatenate([slice1, slice2, slice3, slice4, slice5, slice6], axis=1)



# Sizes of slices
m1 = slice1.shape[1]
m2 = slice2.shape[1]
m3 = slice3.shape[1]
m4 = slice4.shape[1]
m5 = slice5.shape[1]
m6 = slice6.shape[1]

# Create diagonal elements according to the updated rules
diag_entries = np.concatenate([
    np.ones(m1 + m2),            # slice1 + slice2 -> 1
    -np.ones(m3),                # slice3 -> -1
    1j * np.ones(m4),            # slice4 -> i
    -1j * np.ones(m5 + m6)       # slice5 + slice6 -> -i
])

# Create the diagonal matrix
D_0 = np.diag(diag_entries)
# Check if F @ A == A @ D_0 (within numerical tolerance)
is_equal = np.allclose(F @ A, A @ D_0, atol=1e-12)
print("F @ A ≈ A @ D_0 ?", is_equal)
```

#$\theta=N-1$

```python
import numpy as np

m = 2
N = 4 * m
t = N - 1

k = np.arange(N).reshape(N, 1)
l = np.arange(N).reshape(1, N)
w = np.exp(-2j * np.pi / N)
G_c = (1 / np.sqrt(N)) * (w ** ((k - (N - 1) / 2) * (l - (N - 1) / 2)))

# Cosine & Sine matrices
Ct = np.zeros((N, N))
St = np.zeros((N, N))
for k_idx in range(N):
    for l_idx in range(N):
        angle = (k_idx - t / 2) * (l_idx - t / 2) * (2 * np.pi / N)
        Ct[k_idx, l_idx] = (1 / np.sqrt(N)) * np.cos(angle)
        St[k_idx, l_idx] = (1 / np.sqrt(N)) * np.sin(angle)

# X shape
X1 = 0.5 * np.eye(N) + 0.5 * np.fliplr(np.eye(N))
X2 = 0.5 * np.eye(N) - 0.5 * np.fliplr(np.eye(N))

# Matrices
Pc1 = Ct + X1
Pcm1 = X1 - Ct
Pct = St - X2
Pct1 = St + X2

# selected columns
part1 = Pc1[:, m:2 * m]
part2 = Pcm1[:, m:2 * m]
part3 = Pct[:, m:2 * m]
part4 = Pct1[:, m:2 * m]

A_c = np.concatenate([part1, part2, part3, part4], axis=1)

m1 = part1.shape[1]
m2 = part2.shape[1]
m3 = part3.shape[1]
m4 = part4.shape[1]

# Create diagonal elements according to the updated rules
diag_entries = np.concatenate([
    np.ones(m1),
    -np.ones(m2),
    1j * np.ones(m3),
    -1j * np.ones(m4)
])

# Create the diagonal matrix
D_c = np.diag(diag_entries)

# Check if G_c @ A_c == A_c @ D_c (within numerical tolerance)
is_equal = np.allclose(G_c @ A_c, A_c @ D_c, atol=1e-12)
print("G_c @ A_c ≈ A_c @ D_c ?", is_equal)
```
